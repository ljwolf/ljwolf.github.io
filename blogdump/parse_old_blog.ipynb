```python
from datetime import datetime as dt
```

```python
ALL_TAGSTRINGS = []
```

```python
def chunk_file(lines):
    while lines:
        sluice = []
        line = lines.pop(0)
        line = line.strip()
        if line.startswith("Post ID: "):
            postid, date = line.split(', Date')
            date = 'Date' + date
            sluice.extend((postid, date))
            popping = True
            while popping:
                try:
                    line = lines.pop(0)
                    if line.startswith("Post ID: "):
                        lines.insert(0, line)
                        break
                    else:
                        line = line.strip()
                        sluice.append(line)
                except IndexError:
                    break
        yield sluice
def parsepost(chunk):
    headmatter, content = chunk[:7], chunk[7:]
    postid_val = headmatter[0].split(": ")[-1]
    postid = 'id: '+postid_val
    headmatter[0] = postid
    date = headmatter[1].split(": ")[-1]
    date = dt.strptime(date, '%a, %d %b %Y %H:%M:%S').strftime('%Y-%m-%dT%H:%M:%SZ')
    date = 'date: {}'.format(date)
    headmatter[1] = date
    url = headmatter[2].split(": ")[-1]
    urlentry = 'url: ' + url
    headmatter[2] = urlentry
    title = headmatter[-1].split(": ")[-1]
    if title.lower().startswith('title') or title.lower().startswith("link"):
        title = 'Untitled'
    title = title.replace('<p>', '').replace("</p>", '')
    if len(title.split(' ')) > 20:
        content.insert(0, title)
        title = 'Untitled'
    title = 'title: ' + title
    headmatter[-1] = title
    
    content.pop(-1)
    tags = content.pop(-1)
    tags = tags.lower().split(": ")[-1]
    if tags == 'tags:':
        tags = 'tags: ["imported"]'
    else:
        tags = 'tags: ' + '[{}, "imported"]'.format(",".join(['"{}"'.format(s) for s in tags.split(',')])) 
    headmatter.append(tags)
    ALL_TAGSTRINGS.append(tags.lower())
    headmatter = [l for l in headmatter if not l.startswith("Reblog")\
                  or l.startswith("Tags")]
    sourcer = "\n\n\n<small><i> imported from:</i> <a href='{url}'".format(url='https://yetanothergeographer.tumblr.com/{}/{}'.format(postid_val,url))
    sourcer += "<tt>yetanothergeographer</tt></a></small>"
    content.append(sourcer)
    return ('---\n{}\n---\n'.format('\n'.join(headmatter)) + '\n'.join(content), url)
```

```python
def parseimage(chunk):
    headmatter, content = chunk[:7], chunk[7:]
    postid_val = headmatter[0].split(": ")[-1]
    postid = 'id: '+postid_val
    headmatter[0] = postid
    date = headmatter[1].split(": ")[-1]
    date = dt.strptime(date, '%a, %d %b %Y %H:%M:%S').strftime('%Y-%m-%dT%H:%M:%SZ')
    date = 'date: {}'.format(date)
    headmatter[1] = date
    url = headmatter[2].split(": ")[-1]
    if url.lower().startswith('url'):
        del headmatter[2]
    else:
        urlentry = 'url: ' + url
        headmatter[2] = urlentry
    title = headmatter[-1].split(": ")[-1]
    if title.lower().startswith('title') or title.lower().startswith("link"):
        title = 'Untitled'
    title = title.replace('<p>', '').replace("</p>", '')
    if len(title.split(' ')) > 20:
        content.insert(0, title)
        title = 'Untitled'
    title = 'title: ' + title
    headmatter[-1] = title
    
    
    content.pop(-1)
    tags = content.pop(-1)
    tags = tags.lower().split(": ")[-1]
    if tags == 'tags:':
        tags = 'tags: ["imported", "image"]'
    else:
        tags = 'tags: ' + '[{}, "imported", "image"]'.format(",".join(['"{}"'.format(s) for s in tags.split(',')])) 
    headmatter.append(tags)
    ALL_TAGSTRINGS.append(tags.lower())
    headmatter = [l for l in headmatter if not l.startswith("Reblog")\
                  or l.startswith("Tags")]
    sourcer = "\n\n\n<small><i> imported from:</i> <a href='{url}'".format(url='https://yetanothergeographer.tumblr.com/{}/{}'.format(postid_val,url))
    sourcer += "<tt>yetanothergeographer</tt></a></small>"
    content.append(sourcer)
    
    image_template = '<figure> <img src={title}> </figure>\n{content}'
    image = image_template.format(title=title.split(':')[-1].strip(), 
                                  content='\n'.join(content))
    headmatter[-2] = 'title: Untitled'
    return ('---\n{}\n---\n'.format('\n'.join(headmatter)) + image, url)
```
