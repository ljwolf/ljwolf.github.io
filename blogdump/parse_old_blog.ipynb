```python
from datetime import datetime as dt
```

```python
def chunk_file(lines):
    while lines:
        sluice = []
        line = lines.pop(0)
        line = line.strip()
        if line.startswith("Post ID: "):
            postid, date = line.split(', Date')
            date = 'Date' + date
            sluice.extend((postid, date))
            popping = True
            while popping:
                try:
                    line = lines.pop(0)
                    if line.startswith("Post ID: "):
                        lines.insert(0, line)
                        break
                    else:
                        line = line.strip()
                        sluice.append(line)
                except IndexError:
                    break
        yield sluice
def parsepost(chunk):
    headmatter, content = chunk[:7], chunk[7:]
    postid = headmatter[0].split(": ")[-1]
    postid = 'id: '+postid
    headmatter[0] = postid
    date = headmatter[1].split(": ")[-1]
    date = dt.strptime(date, '%a, %d %b %Y %H:%M:%S').strftime('%Y-%m-%dT%H:%M:%SZ')
    date = 'date: {}'.format(date)
    headmatter[1] = date
    url = headmatter[2].split(": ")[-1]
    urlentry = 'url: ' + url
    headmatter[2] = urlentry
    title = headmatter[-1].split(": ")[-1]
    if title.lower().startswith('title') or title.lower().startswith("link"):
        title = 'Untitled'
    title = 'title: '+ title
    headmatter[-1] = title

    tags = content.pop(-1)
    headmatter.append(tags)

    headmatter = [l for l in headmatter if not l.startswith("Reblog")]
    return ('---\n{}---\n'.format('\n'.join(headmatter)) + '\n'.join(content), url)
```

```python
with open("./texts.txt", 'r') as f:
    lines = f.readlines()
    chunks = [parsepost(chunk) for chunk in chunk_file(lines)]
    for chunk,url in chunks:
        print(url)
        with open("./postbypost/{}.md".format(url), 'w') as f:
            f.write(chunk)
with open("./quotes.txt", 'r') as f:
    lines = f.readlines()
    chunks = [parsepost(chunk) for chunk in chunk_file(lines)]
    for chunk,url in chunks:
        print(url)
        with open("./postbypost/{}.md".format(url), 'w') as f:
            f.write(chunk)
with open("./links.txt", 'r') as f:
    lines = f.readlines()
    chunks = [parsepost(chunk) for chunk in chunk_file(lines)]
    for chunk,url in chunks:
        print(url)
        with open("./postbypost/{}.md".format(url), 'w') as f:
            f.write(chunk)
```
